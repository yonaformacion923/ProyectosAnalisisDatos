# Adivina un numero
- Del 1 al 10.

```python
def adivina_numero(minimo = 1, maximo = 10):
    import numpy as np
    num_python = np.random.randint(minimo, maximo+1,1)[0] #Quiero 1 num del 1 al 10)
    num_intentos = 1
    
    while True:
        num_usuario = int(input("Dime un numero del" + str(minimo) + " al " + str(maximo) ))
        if num_python == num_usuario:
            print("Has acertado!! Has tardado ",num_intentos, "intentos")
            break
        else:
            print("Has fallado... intentalo de nuevo...")
            num_intentos += 1
            
    return ({"Numero" : num_python, "Intentos" : num_intentos})
```

```python
adivina_numero()
```

# TIC TAC TOE / 3 EN RAYA

## para 2 jugadores. 1 X o O por turno. Imprime el resultado y equipo ganador

```python
#Planteamos el tablero y lo guardamos en una matriz

import numpy as np
matriz = np.array([[" ", " ", " "],
                     [" ", " ", " "],
                     [" ", " ", " "]])
for fila in matriz:
    print(fila)
```

```python
#limite de tablero. Por cada espacio pon barra vertical
def print_matriz(matriz):   
    print(" -------------")
    for fila in matriz:
        fila_string = " | "
        for elem in fila:
            fila_string += elem + " | "
        print(fila_string)
        print(" -------------")
```

```python
print_matriz(matriz)
```

```python
def anadir_elem(matriz, simboloturno): #fun actualizacion de la matriz (turnos)
    fila, columna = int(input("Que fila?")), int(input("Que columna?"))
    if matriz[fila, columna] == " ": #Si el cajon esta vacio, actualizalo.
         matriz[fila, columna] = simboloturno 
         return matriz #retornamos matriz actualizada.
    else:
         return None
```

```python
anadir_elem(matriz, "X")
```

```python
def victoria(matriz, equipo):
        for fila in matriz: #Bucle para saber si hay 3 simbolos iguales en la fila
             if (fila == equipo).prod():
                 return "Ha ganado el eqipo " + equipo #Match en filas
        for columna in matriz.transpose():
            if (columna == equipo).prod():
                  return "Ha ganado el eqipo " + equipo #Match en Columnas
        if matriz[0,0] == equipo and matriz [1,1] == equipo and matriz[2,2] == equipo: #Match diagonal1
                  return "Ha ganado el eqipo " + equipo
        if matriz[0,2] == equipo and matriz [1,1] == equipo and matriz[2,0] == equipo: #Match diagonal2
                  return "Ha ganado el eqipo " + equipo
        return "Seguir"
```

```python
victoria(matriz, "X")
```

```python
condicion = "Seguir"
matriz = np.array([[" ", " ", " "],
                   [" ", " ", " "],
                   [" ", " ", " "]])

while condicion == "Seguir":
    for team in ["X", "O"]:
        print("Juega el equipo", team)
        print_matriz(matriz)
        matriz = anadir_elem(matriz, team)
        condicion = victoria(matriz, team)
        if condicion != "Seguir":
            print(condicion)
            print_matriz(matriz)
            break 
```

# Me va a tocar la loteria?
- Probabilidades de que te toque la loteria
- tqdm; Muestra barra ejecucion consulta.

```python
import numpy.random as nr
import tqdm

numero = 5 #0000 - 99999 / 1/100.000

resultados = []
for i in tqdm.tqdm(range(10)):
    jugada = 1
    while True:
        if nr.randint(0, 100000) == numero:
            break
        jugada += 1 #Subimos 1 la jugada en cada vuelta.
    resultados.append(jugada) #Guarda 10 jugadas en la lista vacia
```

```python
np.mean(resultados)
```

# CALCULAR ECUACION DE SEGUNDO GRADO
- x = -b +- vb2 - 4ac / 2a

```python
a = 1
b = 0
c = 49
def ecu_2_grado(a=0, b=0,c=0):
    try:
        sol1 = (-b + (b ** 2 - 4*a*c) ** 0.5) / (2*a) # *0.5 es la raiz cuadrada
        sol2 = (-b - (b ** 2 - 4*a*c) ** 0.5) / (2*a)
        return sol1, sol2
    except:
        print("Error o no tiene solucion")
        return None

```

```python
ecu_2_grado(3,-11)
```

# SEPARAR NUMEROS POSITIVOS Y NEGATIVOS
- Logica: recorremos lista de num, si el num >0 anadimos lista positivos.

```python
numeros = [4,5,-2,3,-11,5,-4,3,2,-1]
listapos = []
listaneg = []

for num in numeros:
    try:
        if num > 0:
            listapos.append(num)
        elif num < 0:
            listaneg.append(num)
    except: 
        continue
```

# SUCESION DE FIBONACCI
- 0 1 1 2 3 5 8 13 21...

```python
repeticiones = 64
lista = [0,1]

def fibonacci(repeticiones = 10):
    i = 1
    for repeticion in range(0, repeticiones -2):
        lista.append(lista[i-1] + lista[i])
        i += 1 
    return lista

fibonacci(64)
```

# TABLAS DE MULTIPLICAR
- Logica: andidar 2 for: 1 lo multiplica por el [2]

```python

def tablas_multi(maximo = 10):
    numeros = list(range(1,maximo+1))
    for num1 in numeros:
        print("\n Tabla del", num1, "\n")
        for num2 in numeros:
            print(num1, "x" , num2, "=", num1*num2)

tablas_multi(10)
```

# CALCULAR LA MEDIANA
- Preparamos mediana de num par e impar.
- Ordenamos lista

```python
def mediana(datos):
    ordenada = sorted(datos)
    if len(ordenada) % 2 == 0:
        return (ordenada[int(len(ordenada)/2 - 1)] + ordenada[int(len(ordenada)/2)] )/2
    else: 
        return ordenada[int(len(ordenada)/2)]
```

```python
mediana([2,3,1,4,4,1,2,4,5,32,21,2])
```
